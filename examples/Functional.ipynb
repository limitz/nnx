```python
import os
import torch
import torch.nn as nn
import torch.nn.functional as F
import nnx
import nnx.functional as Fx
import nnx.console as console
import time
```

```python
files = {"a.test": torch.rand(256), 
         "b.test": torch.rand(128), 
         "d.test": torch.rand(127),
         "c.test": torch.rand(512)}

for path, content in reversed(files.items()):
    torch.save(content, "/tmp/" + path)
    time.sleep(1)


console.print("<cyan><b>Sort by name</>")
for path in Fx.glob("/tmp/*.test", order_by="name"):
    console.print(" - ", path, 
                  f"<gray>modified: <green>{os.path.getmtime(path):8.02f}</>", 
                  f"<gray>filesize: <lime>{os.path.getsize(path):3d}</>")
console.print()

console.print("<cyan><b>Sort by modification time</>")
for path in Fx.glob("/tmp/*.test", order_by="modified"):
    console.print(" - ", path, 
                  f"<gray>modified: <green>{os.path.getmtime(path):8.02f}</>", 
                  f"<gray>filesize: <lime>{os.path.getsize(path):3d}</>")
console.print()

console.print("<cyan><b>Sort by modification time descending</>")
for path in Fx.glob("/tmp/*.test", order_by="modification time desc"):
    console.print(" - ", path, 
                  f"<gray>modified: <green>{os.path.getmtime(path):8.02f}</>", 
                  f"<gray>filesize: <green>{os.path.getsize(path):3d}</>")
console.print()

console.print("<cyan><b>Sort by file size</>")
for path in Fx.glob("/tmp/*.test", order_by="size"):
    console.print(" - ", path, 
                  f"<gray>modified: <green>{os.path.getmtime(path):8.02f}</>", 
                  f"<gray>filesize: <green>{os.path.getsize(path):3d}</>")
```
